```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from keras.models import Sequential
from keras.layers import Conv2D, MaxPooling2D
from keras.layers import Activation, Dropout, Flatten, Dense
from keras import backend as K
```

```python
img_width, img_height = 150, 150
train_data_dir = '/content/drive/MyDrive/v_data/test'
validation_data_dir = '/content/drive/MyDrive/v_data/train'
```

```python
nb_train_samples = 400
nb_validation_samples = 100
epochs = 50
batch_size =15
```

```python
if K.image_data_format== 'channels_first':
  input_shape = (3, img_width, img_height)
else:
  input_shape = (img_width, img_height, 3)
```

```python
from keras.layers import Input, LeakyReLU

model=Sequential()
model.add(Input(shape=input_shape))
model.add(Conv2D(32,(3,3),activation='relu'))
model.add(Conv2D(64,(3,3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2)))
model.add(Dropout(0.25))
model.add(LeakyReLU(negative_slope=0.2))

model.add(Flatten())
model.add(Dense(128,activation='relu'))
model.add(Dense(1,activation='sigmoid')) # Changed to 1 unit and sigmoid activation for binary classification
```

```python
#compile the CNN
model.compile(optimizer='adam',
              loss = 'binary_crossentropy', metrics=['accuracy']) # Changed loss to binary_crossentropy
```

```python
#Data Generator

train_datagen = ImageDataGenerator(
    rescale=1./255,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True
)

validation_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    train_data_dir,
    target_size=(img_width, img_height),
    batch_size=batch_size,
    class_mode='binary'
)

validation_generator = validation_datagen.flow_from_directory(
    validation_data_dir,
    target_size=(img_width, img_height),
    batch_size=batch_size,
    class_mode='binary'
)

model.fit(
    train_generator,
    steps_per_epoch=nb_train_samples // batch_size,
    epochs=epochs,
    validation_data=validation_generator,
    validation_steps=nb_validation_samples // batch_size
)
```

```python
model.save('model.keras')
```

```python
from keras.models import load_model
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np

model = load_model ('model.h5') # Changed from 'model.keras' to 'model.h5'

image = load_img('/content/drive/MyDrive/v_data/car.jpg',target_size= (150,150))
input_arr = img_to_array(image)
input_arr = input_arr/255.0 # Corrected variable name from 'img' to 'input_arr'
input_arr = np.expand_dims(input_arr, axis=0) # Corrected variable name from 'img' to 'input_arr'
label = model.predict(input_arr)
print(label)
```

```python
img = np.array(image)
img = img / 255.0
img = np.expand_dims(img, axis=0)
label = model.predict(img)
print("Predicted class (0 - Cars , 1- Planes):", label[0,0].round())
```
